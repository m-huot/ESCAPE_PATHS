```python
from Bio import SeqIO
from Bio.SeqRecord import SeqRecord

# Input and output file paths
input_file = "bloom_deep_scan_ACE2_with_sequence.fasta"
output_file = "bloom_esmif.fasta"

# Load, trim, and store sequences
trimmed_records = []
for record in SeqIO.parse(input_file, "fasta"):
    trimmed_seq = record.seq[2:-5]  # Remove first 2 and last 5 residues
    trimmed_record = SeqRecord(trimmed_seq, id=record.id, description=record.description)
    trimmed_records.append(trimmed_record)

# Save to new FASTA
SeqIO.write(trimmed_records, output_file, "fasta")
```
