# Analyse the fitted concentrations of antibodies

```python
import pandas as pd
import numpy as np


```

```python
ab_class_df=pd.read_csv('all_ab_escape.csv')
#keep condition_type=antibody    
ab_class_df = ab_class_df[ab_class_df['condition_type']=='antibody']       
#keep lab=Bloom_JD
ab_class_df = ab_class_df[ab_class_df['lab']=='Bloom_JD']           
#for each condition, find the associated condition_subtype. create a df with condition, condition_subtype
conditions = ab_class_df['condition'].unique()
print(conditions)
condition_subtype = []
for condition in conditions:
    condition_subtype.append(ab_class_df[ab_class_df['condition']==condition]['condition_subtype'].values[0])
    print(condition, condition_subtype)
condition_subtype_df = pd.DataFrame({'condition':conditions, 'condition_subtype':condition_subtype})
```
