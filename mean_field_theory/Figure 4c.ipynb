# Viz using pymol software called from python

Plot entropy of sites


```python
import numpy as np
import pandas as pd
import pymol
from pymol import cmd



residues=[357, 363, 365, 372, 378, 385, 394, 417, 450, 487, 498, 518, 519]

site_numbers = [
    332, 339, 346, 356, 368, 371, 373, 376, 403, 405, 408, 417, 440, 444, 446, 447, 449, 450,
    452, 455, 456, 460, 471, 478, 484, 486, 490, 493, 496, 498, 501, 505
]

pymol.finish_launching()

pdb_id = "6M0J"
cmd.fetch(pdb_id, pdb_id)

cmd.bg_color("white")
cmd.hide("everything", "all")
cmd.select("Spike", f"{pdb_id} and chain E")
cmd.select("ACE2", f"{pdb_id} and chain A")
cmd.show("cartoon", "Spike")
cmd.show("cartoon", "ACE2")
cmd.color("blue", "ACE2")
cmd.set("cartoon_transparency", 0.1, "ACE2")
cmd.color("cyan", "Spike")


  


highlighted = [resi for resi in residues
               if resi in site_numbers ]

for resi in highlighted:
    sel = f"{pdb_id} and chain E and resi {resi}"
    cmd.show("spheres", sel)
    cmd.set("sphere_scale", 0.7, sel)
    cmd.color("hotpink", sel)  # Optional override for visibility

not_highlighted = [resi for resi in residues
               if resi not in site_numbers ]
for resi in not_highlighted:
    sel = f"{pdb_id} and chain E and resi {resi}"

    cmd.color("red", sel)  # Optional override for visibility
cmd.zoom("Spike", buffer=20)
cmd.set("ray_opaque_background", 0)

cmd.rotate("y", 0) 

cmd.rotate("x", 10)
cmd.ray(1920, 1080)

```
